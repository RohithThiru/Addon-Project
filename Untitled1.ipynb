```python
# Cell 1: imports
import os
import io
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist, cifar10
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from tensorflow.keras.preprocessing.image import array_to_img
from google.colab import files

print("TensorFlow version:", tf.__version__)
```

```python
# Cell 2: choose dataset - uncomment one block

DATASET = "MNIST"   # "MNIST" or "CIFAR10"

if DATASET == "MNIST":
    (x_train, _), (x_test, _) = mnist.load_data()
    # expand dims -> (N, H, W, 1)
    x_train = np.expand_dims(x_train, -1).astype("float32") / 255.0
    x_test = np.expand_dims(x_test, -1).astype("float32") / 255.0
    IMG_SHAPE = x_train.shape[1:]
    print("Using MNIST", IMG_SHAPE, "train samples:", len(x_train))
else:
    (x_train, _), (x_test, _) = cifar10.load_data()
    x_train = x_train.astype("float32") / 255.0
    x_test = x_test.astype("float32") / 255.0
    IMG_SHAPE = x_train.shape[1:]
    print("Using CIFAR-10", IMG_SHAPE, "train samples:", len(x_train))
```

```python
# Cell 3: noise functions (Gaussian + Salt-and-Pepper)
import random

def add_gaussian_noise(images, sigma=0.2):
    noisy = images + np.random.normal(loc=0.0, scale=sigma, size=images.shape)
    noisy = np.clip(noisy, 0., 1.)
    return noisy

def add_salt_and_pepper_noise(images, amount=0.05, s_vs_p=0.5):
    noisy = images.copy()
    N = images.shape[0]
    H, W = images.shape[1], images.shape[2]
    C = 1 if images.ndim == 4 and images.shape[3] == 1 else images.shape[3] if images.ndim==4 else 1
    num_salt = np.ceil(amount * H * W * s_vs_p)
    num_pepper = np.ceil(amount * H * W * (1. - s_vs_p))
    for i in range(N):
        # Salt
        coords = [np.random.randint(0, H, int(num_salt)), np.random.randint(0, W, int(num_salt))]
        if C == 1:
            noisy[i, coords[0], coords[1], 0] = 1
        else:
            noisy[i, coords[0], coords[1], :] = 1
        # Pepper
        coords = [np.random.randint(0, H, int(num_pepper)), np.random.randint(0, W, int(num_pepper))]
        if C == 1:
            noisy[i, coords[0], coords[1], 0] = 0
        else:
            noisy[i, coords[0], coords[1], :] = 0
    return noisy
```

```python
# Cell 4: prepare training/validation sets (use subset for speed)
NUM_TRAIN = 20000 if DATASET=="CIFAR10" else 60000  # full defaults; Colab free GPU may handle it
# To speed up demo/testing, you can reduce:
train_sample = 20000 if DATASET=="CIFAR10" else 10000

x_train = x_train[:train_sample]
x_test = x_test[:2000]  # keep test small for quick runs

# create noisy versions (choose one)
noisy_train = add_gaussian_noise(x_train, sigma=0.2)
noisy_test = add_gaussian_noise(x_test, sigma=0.2)

# also build an alternative dataset with salt-and-pepper (if you want)
noisy_train_sp = add_salt_and_pepper_noise(x_train, amount=0.05)
noisy_test_sp = add_salt_and_pepper_noise(x_test, amount=0.05)

print("Shapes:", x_train.shape, noisy_train.shape)
```

```python
# Cell 5: build a small convolutional denoising autoencoder
def build_autoencoder(input_shape):
    inp = layers.Input(shape=input_shape)
    x = inp
    # Encoder
    x = layers.Conv2D(32, 3, activation="relu", padding="same")(x)
    x = layers.MaxPool2D(2, padding="same")(x)
    x = layers.Conv2D(64, 3, activation="relu", padding="same")(x)
    x = layers.MaxPool2D(2, padding="same")(x)
    # Bottleneck
    x = layers.Conv2D(128, 3, activation="relu", padding="same")(x)
    # Decoder
    x = layers.UpSampling2D(2)(x)
    x = layers.Conv2D(64, 3, activation="relu", padding="same")(x)
    x = layers.UpSampling2D(2)(x)
    x = layers.Conv2D(32, 3, activation="relu", padding="same")(x)
    # Output: same channels as input
    out_channels = input_shape[2] if len(input_shape)==3 else 1
    out = layers.Conv2D(out_channels, 3, activation="sigmoid", padding="same")(x)
    model = models.Model(inp, out, name="denoising_autoencoder")
    return model

autoencoder = build_autoencoder(IMG_SHAPE)
autoencoder.compile(optimizer="adam", loss="mse")
autoencoder.summary()
```

```python
# Cell 6: train the model (using Gaussian noisy -> clean)
EPOCHS = 25
BATCH_SIZE = 128

# Use noisy_train (Gaussian). To train on salt-and-pepper use noisy_train_sp instead.
history = autoencoder.fit(
    noisy_train, x_train,
    validation_data=(noisy_test, x_test),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=[EarlyStopping(monitor="val_loss", patience=4, restore_best_weights=True)]
)
```

```python
def show_images_grid(orig, noisy, recon, n=8, figsize=(16,6), title=""):
    plt.figure(figsize=figsize)
    for i in range(n):
        # original
        plt.subplot(3, n, i+1)
        img = orig[i]
        plt.axis('off')
        if img.shape[-1]==1:
            plt.imshow(img.squeeze(), cmap='gray')
        else:
            plt.imshow(img)
        if i==0: plt.ylabel("Original")
        # noisy
        plt.subplot(3, n, n + i+1)
        img = noisy[i]
        plt.axis('off')
        if img.shape[-1]==1:
            plt.imshow(img.squeeze(), cmap='gray')
        else:
            plt.imshow(img)
        if i==0: plt.ylabel("Noisy")

        plt.subplot(3, n, 2*n + i+1)
        img = recon[i]
        plt.axis('off')
        if img.shape[-1]==1:
            plt.imshow(img.squeeze(), cmap='gray')
        else:
            plt.imshow(img)
        if i==0: plt.ylabel("Reconstructed")
    plt.suptitle(title)
    plt.tight_layout()
    plt.show()

# generate reconstructions
recon_test = autoencoder.predict(noisy_test, batch_size=128)
# show a small grid
show_images_grid(x_test, noisy_test, recon_test, n=8, figsize=(16,6), title="Gaussian noise denoising")
```
